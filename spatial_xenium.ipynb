```python
# Load and prepare anndata object
# Add the metadata column
# Seurat Cluster co-ordinates
seurat_cord = pd.read_csv('/Users/other/Xenium_Napari_LTF_Masking/seurat_cord.csv')


from spatialdata import polygon_query
import anndata as ad
from collections import defaultdict


def create_anndata(sdata):
    filtered_tables_unmerged: dict[str, list[ad.AnnData]] = defaultdict(list)
    filtered_tables: dict[str, ad.AnnData] = {}

    for shape in ["LTF_SMG"]:
        for polygon in sdata[shape].geometry:
            table = polygon_query(sdata, polygon=polygon, target_coordinate_system="global")["table"]
            filtered_tables_unmerged[shape].append(table)
    filtered_tables[shape] = ad.concat(filtered_tables_unmerged[shape])
    anndata = filtered_tables['LTF_SMG']
    anndata=anndata[anndata.obs['cell_id'].isin(list(seurat_cord['cell_id']))]
    return anndata
    
```

```python
# BOS samples
BOS11A_h5 = create_anndata(BOS11A_SMG)

BOS12D_h5 = create_anndata(BOS12D_SMG)

BOS13E_h5 = create_anndata(BOS13E_SMG)
```

```python
# Merge All objects
lis=[ BOS11A_h5, BOS12D_h5, BOS13E_h5]
bos_final_smg=sp.concat(lis, join='outer')
bos_final_smg
```

```python
# from spatialdata.query import polygon_query
from shapely import union_all

def get_smg_region_spatialdata(sdata_obj):
    # 1. Extract & fix geometries
    fixed_geoms = sdata_obj['LTF_SMG'].geometry.buffer(0)
    
    # 2. Merge into a single valid geometry
    combined_region = union_all(fixed_geoms)
    
    # 3. Query (table argument removed - handled automatically)
    sdata_subset = polygon_query(
        sdata_obj,
        polygon=combined_region,
        target_coordinate_system="global"
    )
    
    return sdata_subset

# Usage
sdata_cut_BOS11A = get_smg_region_spatialdata(BOS11A_SMG)
sdata_cut_BOS12D = get_smg_region_spatialdata(BOS12D_SMG)
sdata_cut_BOS13E = get_smg_region_spatialdata(BOS13E_SMG)
```

```python
def generate_spatial_image(sd_obj, file):
    fig, ax = plt.subplots(figsize=(6,5))
    sd_obj["cell_boundaries"].geometry.plot(
        ax=ax,
        facecolor="none",
        edgecolor="orange",
        linewidth=1,
        alpha=1)

    ax.set_axis_off()

    plt.tight_layout()

    # Save as PDF
    plt.savefig(file,
                 format="pdf",
        bbox_inches="tight"
    )

    plt.show()

generate_spatial_image(sdata_cut_BOS11A, '/Users/other/Xenium_Napari_LTF_Masking/BOS/BOS11A_LTF_SMG.pdf')
generate_spatial_image(sdata_cut_BOS12D, '/Users/other/Xenium_Napari_LTF_Masking/BOS/BOS12D_LTF_SMG.pdf')
generate_spatial_image(sdata_cut_BOS13E, '/Users/other/Xenium_Napari_LTF_Masking/BOS/BOS13E_LTF_SMG.pdf')
```
